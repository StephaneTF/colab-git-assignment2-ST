# Task 9

```python
# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.

# Your code here:
# Iterator
class EvenIterator:
    def __init__(self, count):
        self.max_count = count
        self.current_count = 0
        self.current_even = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current_count >= self.max_count:
            raise StopIteration

        result = self.current_even
        self.current_even += 2
        self.current_count += 1
        return result

# Using the Iterator
print("--- Iterator Output ---")
even_iter = EvenIterator(5)
for num in even_iter:
    print(num)
print()


# Generator
def even_generator(count):
    current_even = 0
    for _ in range(count):
        yield current_even
        current_even += 2

# Using the Generator
print("--- Generator Output ---")
even_gen = even_generator(5)
for num in even_gen:
    print(num)
```

# Task 10

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:
numbers = [1, 2, 3, 4, 5]
# Use map to square all numbers in a list
squared_numbers = list(map(lambda x: x ** 2, numbers))

# Use reduce to find the product of all numbers in a list
product_of_numbers = reduce(lambda x, y: x * y, numbers)

# Use filter to get only even numbers from a list
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
# Displaying the results
print(f"Original numbers: {numbers}")
print(f"Squared (Map):    {squared_numbers}")
print(f"Product (Reduce): {product_of_numbers}")
print(f"Evens (Filter):   {even_numbers}")
```

# Task 11

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:
class Rectangle:
    def __init__(self, length, width):
        # Initializing the attributes
        self.length = length
        self.width = width

    def calculate_area(self):
        # Area = length * width
        return self.length * self.width

    def calculate_perimeter(self):
        # Perimeter = 2 * (length + width)
        return 2 * (self.length + self.width)

# Verification and Testing
# Creating an instance of Rectangle with length=5 and width=3
rect = Rectangle(5, 3)

print(f"Rectangle Dimensions: Length = {rect.length}, Width = {rect.width}")
print(f"Calculated Area:       {rect.calculate_area()}")
print(f"Calculated Perimeter:  {rect.calculate_perimeter()}")
```
